```python
import os
import openai

# Setzen Sie Ihren OpenAI-API-Key
openai.api_key = "sk-..."

def get_code_quality_feedback(file_path, code):
    prompt = f"Bewerte die Codequalität der folgenden Datei und gib Verbesserungsvorschläge:\n\nDatei: {file_path}\n\n{code}\n"
    response = openai.ChatCompletion.create(
        model="gpt-4",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=500
    )
    return response.choices[0].message.content.strip()

def scan_directory(directory):
    results = {}
    for root, _, files in os.walk(directory):
        for file in files:
            if file.endswith(('.py', '.js', '.ts', '.java', '.cpp', '.c', '.cs')):  # beliebig anpassen
                file_path = os.path.join(root, file)
                with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
                    code = f.read()
                feedback = get_code_quality_feedback(file_path, code)
                results[file_path] = feedback
    return results

if __name__ == "__main__":
    directory = "."  # aktuelles Verzeichnis
    feedbacks = scan_directory(directory)
    for file, feedback in feedbacks.items():
        print(f"--- {file} ---\n{feedback}\n")
```
