```python
from scipy.spatial import distance

# Definiere die Punkte
A = (1, 1)
B = (1, 2)
C = (2, 1)
D = (5, 4)
E = (6, 5)
F = (5, 5)


z1 = A
z2 = D
```

```python
# Berechne die euklidischen Distanzen
# d(A, z1)
d1 = distance.euclidean(A, z1)
print("A:", d1)

# d(A, z2)
d2 = distance.euclidean(A, z2)
print("A:", d2)

# d(B, z1)
d3 = distance.euclidean(B, z1)
print("B:", d3)

# d(B, z2)
d4 = distance.euclidean(B, z2)
print("B:", d4)

# d(C, z1)
d5 = distance.euclidean(C, z1)
print("C:", d5)

# d(C, z2)
d6 = distance.euclidean(C, z2)
print("C:", d6)

# d(D, z1)
d7 = distance.euclidean(D, z1)
print("D:", d7)

# d(D, z2)
d8 = distance.euclidean(D, z2)
print("D:", d8)

# d(E, z1)
d9 = distance.euclidean(E, z1)
print("E:", d9)

# d(E, z2)
d10 = distance.euclidean(E, z2)
print("E:", d10)

# d(F, z1)
d11 = distance.euclidean(F, z1)
print("F:", d11)

# d(F, z2)
d12 = distance.euclidean(F, z2)
print("F:", d12)
```

```python
# Neue Zentren berechnen (Mittelwerte)
# Zentrum z1 für Punkte A, B, C
z1x = (1 + 1 + 2) / 3
z1y = (1 + 2 + 1) / 3

# Zentrum z2 für Punkte D, E, F
z2x = (5 + 6 + 5) / 3
z2y = (4 + 5 + 5) / 3

print(f"Neues Zentrum z1: ({z1x}, {z1y})")
print(f"Neues Zentrum z2: ({z2x}, {z2y})")
```

```python
# Skizziere die Punkte und Zentren
import matplotlib.pyplot as plt
points = [A, B, C, D, E, F]
z1 = (z1x, z1y)
z2 = (z2x, z2y)
plt.scatter(*zip(*points), color='blue', label='Datenpunkte')
plt.scatter(*z1, color='red', label='Zentrum z1')
plt.scatter(*z2, color='green', label='Zentrum z2')
plt.legend()
plt.title('K-Means Clustering')
plt.xlabel('X-Achse')
plt.ylabel('Y-Achse')
plt.grid()
plt.show()
```

```python
from scipy.spatial import distance

points = {
    "A": (1, 1),
    "B": (1, 2),
    "C": (2, 1),
    "D": (5, 4),
    "E": (6, 5),
    "F": (5, 5),
}

zentren = {
    "z_1": (1, 1),
    "z_2": (5, 4),
}

print(distance.euclidean(points["A"], zentren["z_1"]))

for bezeichner, p in points.items():
    d1 = distance.euclidean(p, zentren["z_1"])
    d2 = distance.euclidean(p, zentren["z_2"])
    knearest = "z_1" if d1 <= d2 else "z_2"
    print(f"{bezeichner}: d({bezeichner}, z_1)={d1:.4f}, d({bezeichner}, z_2)={d2:.4f} -> zu {knearest}")
```
