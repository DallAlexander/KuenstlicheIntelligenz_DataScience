```python
import numpy as np
import matplotlib.pyplot as plt
```

# Aufgabe Verteilungen

Referenz-Diagramme: http://www.mesosworld.ch/lerninhalte/Univ_Beschreib/de/html/unit_VerteilungsFormen.html

```python
# Zufallsdaten generieren
rng = np.random.default_rng(42) # Reproduzierbarkeit durch festen Seed 42
n = 10000 # Anzahl der erzeugten Datenpunkte
```

### 1) Gleichverteilung

```python
# U(0,1) = Werte zwischen 0 und 1
gleichverteilung_uniform = rng.uniform(0.0, 1.0, n)

# Plot der Gleichverteilung
plt.figure(figsize=(8, 5))
plt.hist(gleichverteilung_uniform, bins=50, density=True, color="#4C78A8", alpha=0.8, edgecolor="white")
plt.title("Gleichverteilung U(0,1)")
plt.xlabel("Wert")
plt.ylabel("Dichte")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 2) Normalverteilung

```python
# N(µ=0, σ=1) – unimodal, symmetrisch
# Zahlen mit Mittelwert 0 und Standardabweichung 1 (Gaußsche Glockenkurve)
normalverteilung = rng.normal(loc=0.0, scale=1.0, size=n)

# Plot der Normalverteilung
plt.figure(figsize=(8, 5))
plt.hist(normalverteilung, bins=50, density=True, color="#4C78A8", alpha=0.8, edgecolor="white")
plt.title("Normalverteilung")
plt.xlabel("Wert")
plt.ylabel("Dichte")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 3) Unimodal linksschief

```python
# (Beta(5,2) auf [0,1])
# Beta-Verteilung mit a=5, b=2 → linksschiefe Verteilung
unimodal_linksschief = rng.beta(a=5.0, b=2.0, size=n)

# Plot der Verteilung unimodal linksschief
plt.figure(figsize=(8, 5))
plt.hist(unimodal_linksschief, bins=50, density=True, color="#4C78A8", alpha=0.8, edgecolor="white")
plt.title("Unimodal linksschief")
plt.xlabel("Wert")
plt.ylabel("Dichte")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 4) Unimodal rechtsschief

```python
# (Beta(2,5) auf [0,1])
# Beta-Verteilung mit a=2, b=5 → rechtsschiefe Verteilung
# linksschief umgekehrt
unimodal_rechtsschief = rng.beta(a=2.0, b=5.0, size=n)

# Plot der Verteilung unimodal rechtsschief
plt.figure(figsize=(8, 5))
plt.hist(unimodal_rechtsschief, bins=50, density=True, color="#4C78A8", alpha=0.8, edgecolor="white")
plt.title("Unimodal rechtsschief")
plt.xlabel("Wert")
plt.ylabel("Dichte")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5) Exponentialverteilung

```python
# 5) Exponentialverteilung (λ=1 → scale=1)
# Viele kleine Werte, wenige große Werte
exponentialverteilung = rng.exponential(scale=1.0, size=n)

# Plot der Exponentialverteilung
plt.figure(figsize=(8, 5))
plt.hist(exponentialverteilung, bins=50, density=True, color="#4C78A8", alpha=0.8, edgecolor="white")
plt.title("Exponentialverteilung")
plt.xlabel("Wert")
plt.ylabel("Dichte")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 6) Multimodale Verteilung

```python
mix_p = 0.5 # 50 % der Werte zufällig ausgewählt aus beiden Normalverteilungen
mask = rng.random(n) < mix_p # Maske für Auswahl der Verteilung
multimodale_verteilung = np.empty(n)
multimodale_verteilung[mask] = rng.normal(loc=-2.0, scale=0.5, size=mask.sum()) # 50 % aus NV mit Mittelwert -2
multimodale_verteilung[~mask] = rng.normal(loc=+2.0, scale=0.8, size=(~mask).sum()) # 50 % aus NV mit Mittelwert +2

# Plot der multimodalen Verteilung
plt.figure(figsize=(8, 5))
plt.hist(multimodale_verteilung, bins=50, density=True, color="#4C78A8", alpha=0.8, edgecolor="white")
plt.title("Multimodale Verteilung")
plt.xlabel("Wert")
plt.ylabel("Dichte")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
