```python
import pandas as pd
import xml.etree.ElementTree as ET
```

# Aufgabe 2: Semistrukturierte Daten

### JSON einlesen und normalisieren

```python
file_path = 'data/semistrukturiert/Personen.json'

df = pd.read_json(file_path, encoding='utf-8')
df = pd.json_normalize(df['personen'])

df.info()
```

```python
df.describe()
```

#### Inkorrekte Werte in Alter-Spalte finden

```python
print(f"Datentyp Spalte Alter: {df['alter'].dtype}")
df
```

```python
# Median berechnen (vorher sicherstellen, dass numerisch)
df['alter'] = pd.to_numeric(df['alter'], errors='coerce')
median_alter = df['alter'].median()
print(f"Median Alter: {median_alter}")

# Variante 1: Zuordnung ohne inplace (empfohlen)
df['alter'] = df['alter'].fillna(median_alter)

# Alternative Variante 2: Auf DataFrame-Ebene inplace
# df.fillna({'alter': median_alter}, inplace=True)

print(df)
print(f"Datentyp Spalte Alter: {df['alter'].dtype}")
```

### XML einlesen und normalisieren

```python
file_path = 'data/semistrukturiert/Personen.xml'

tree = ET.parse(file_path)
root = tree.getroot()

data = []
for person in root.findall('person'):
    eintrag = {}
    for attribute in person:
        eintrag[attribute.tag] = attribute.text
    data.append(eintrag)
df_xml = pd.DataFrame(data)

df_xml.info()
```
