```python
import pandas as pd
import matplotlib as plt
```

# Aufgabe 1: Strukturierte Daten

```python
df = pd.read_csv('data/strukturiert/Personen.csv', delimiter=';')
df.info()
```

### Lageparameter

```python
df.describe()
# Durchschnitt berechnen
durchschnitt = df['Alter'].mean()
print(f'Durchschnitt: {durchschnitt}')

# Median berechnen
median = df['Alter'].median()
print(f'Median: {median}')

# Modus berechnen
modus = df['Alter'].mode()[0]
print(f'Modus: {modus}')

# Quantile berechnen
Q1 = df['Alter'].quantile(0.25)
Q2 = df['Alter'].quantile(0.5)
Q3 = df['Alter'].quantile(0.75)
print(f'Q1: {Q1}, Q2: {Q2}, Q3: {Q3}')
```

## Streuungsmaße

```python
# Spannweite berechnen
spannweite = df['Alter'].max() - df['Alter'].min()
print(f'Spannweite: {spannweite}')

# Interquartialsabstand berechnen
Q1 = df['Alter'].quantile(0.25)
Q3 = df['Alter'].quantile(0.75)
IQR = Q3 - Q1
print(f'Interquartialsabstand (IQR): {IQR}')

#Varianz berechnen
varianz = df['Alter'].var()
print(f'Varianz: {varianz}')

#Standardabweichung berechnen
standardabweichung = df['Alter'].std()
print(f'Standardabweichung: {standardabweichung}')
```

```python
df['Alter'].plot.box()
plt.pyplot.title('Boxplot: Alter')
plt.pyplot.ylabel('Alter')
plt.pyplot.show()
```
