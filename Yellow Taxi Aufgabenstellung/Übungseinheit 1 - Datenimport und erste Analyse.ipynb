### Fehlende Werte

#### Ausreißer

##### Methode 1: IQR-Methode (Interquartilsabstand)

```python
# Funktion zur Identifikation von Ausreißern mit IQR-Methode
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return outliers, lower_bound, upper_bound

# Numerische Spalten für Ausreißer-Analyse
numeric_columns = ['passenger_count', 'trip_distance', 'fare_amount', 'tip_amount',
                   'tolls_amount', 'total_amount', 'extra', 'mta_tax',
                   'improvement_surcharge', 'congestion_surcharge', 'airport_fee']

print("Ausreißer-Analyse (IQR-Methode):")
print("="*80)
for col in numeric_columns:
    if col in df.columns:
        outliers, lower, upper = detect_outliers_iqr(df, col)
        outlier_percent = (len(outliers) / len(df)) * 100
        print(f"\n{col}:")
        print(f"  Untere Grenze: {lower:.2f}")
        print(f"  Obere Grenze: {upper:.2f}")
        print(f"  Anzahl Ausreißer: {len(outliers):,} ({outlier_percent:.2f}%)")
        if len(outliers) > 0:
            print(f"  Min-Wert (Ausreißer): {df[col].min():.2f}")
            print(f"  Max-Wert (Ausreißer): {df[col].max():.2f}")
```

##### Methode 2: Z-Score (für normalverteilte Daten)

```python
from scipy import stats
import numpy as np

# Z-Score Methode (Werte mit |z| > 3 sind Ausreißer)
print("Ausreißer-Analyse (Z-Score > 3):")
print("="*80)
for col in ['fare_amount', 'trip_distance', 'tip_amount', 'total_amount']:
    if col in df.columns:
        z_scores = np.abs(stats.zscore(df[col].dropna()))
        outliers_z = len(z_scores[z_scores > 3])
        outlier_percent_z = (outliers_z / len(df)) * 100
        print(f"\n{col}:")
        print(f"  Anzahl Ausreißer (|Z| > 3): {outliers_z:,} ({outlier_percent_z:.2f}%)")
```

##### Spezifische Probleme identifizieren

```python
print("Spezifische Datenprobleme:")

# Negative Werte bei Zahlenwerten
print("\n1. Negative Werte:")
for col in ['fare_amount', 'tip_amount', 'tolls_amount', 'total_amount', 'trip_distance']:
    if col in df.columns:
        negativ_counter = (df[col] < 0).sum()
        print(f"  {col}: {negativ_counter:,} negative Werte")

# Null-Werte bei relevanten Attributen
print("\n2. Null-Werte bei relevanten Attributen:")
for col in ['passenger_count', 'trip_distance', 'fare_amount', 'total_amount']:
    if col in df.columns:
        null_counter = (df[col] == 0).sum()
        null_anteilig = (null_counter / len(df)) * 100
        print(f"  {col}: {null_counter:,} Null-Werte ({null_anteilig:.2f}%)")

# Unrealistische Werte
print("\n3. Unrealistische Werte:")
if 'passenger_count' in df.columns:
    anzahl_mitfahrer = (df['passenger_count'] > 6).sum()
    print(f"  Mehr als 6 Passagiere: {anzahl_mitfahrer:,}")
    null_mitfahrer = (df['passenger_count'] == 0).sum()
    print(f"  Null Passagiere: {null_mitfahrer:,}")

if 'trip_distance' in df.columns:
    sehr_lange_fahrten = (df['trip_distance'] > 100).sum()
    print(f"  Fahrten > 100 Meilen: {sehr_lange_fahrten:,}")

if 'fare_amount' in df.columns:
    sehr_hoher_preis = (df['fare_amount'] > 750).sum()
    print(f"  Fahrpreis > $750: {sehr_hoher_preis:,}")
    

# Zeitliche Probleme
print("\n4. Zeitliche Probleme:")
df['reisedauer'] = (df['tpep_dropoff_datetime'] - df['tpep_pickup_datetime']).dt.total_seconds() / 60
negative_dauer = (df['reisedauer'] < 0).sum()
null_dauer = (df['reisedauer'] == 0).sum()
sehr_lange_reise = (df['reisedauer'] > 180).sum()  # > 3 Stunden
print(f"  Negative Fahrtdauer: {negative_dauer:,}")
print(f"  Null Minuten Fahrtdauer: {null_dauer:,}")
print(f"  Fahrtdauer > 3 Stunden: {sehr_lange_reise:,}")

# Inkonsistenzen
print("\n5. Inkonsistenzen:")
if 'trip_distance' in df.columns and 'fare_amount' in df.columns:
    # Fahrten ohne Distanz aber mit Fahrpreis
    keine_distanz_mit_preis = ((df['trip_distance'] == 0) & (df['fare_amount'] > 0)).sum()
    print(f"  Keine Distanz aber Fahrpreis > 0: {keine_distanz_mit_preis:,}")

    # Lange Fahrten mit sehr niedrigem Preis
    lange_fahrt_günstig = ((df['trip_distance'] > 10) & (df['fare_amount'] < 5)).sum()
    print(f"  Lange Fahrt (>10 Meilen) mit Preis <$5: {lange_fahrt_günstig:,}")
```

##### Zusammenfassung der Ausreißer

```python
print("\n" + "="*80)
print("ZUSAMMENFASSUNG DER DATENQUALITÄTSPROBLEME")
print("="*80)

total_rows = len(df)
problematic_rows = set()

# Sammle alle problematischen Zeilen
if 'fare_amount' in df.columns:
    problematic_rows.update(df[df['fare_amount'] < 0].index)
    problematic_rows.update(df[df['fare_amount'] > 500].index)
if 'trip_distance' in df.columns:
    problematic_rows.update(df[df['trip_distance'] < 0].index)
    problematic_rows.update(df[df['trip_distance'] > 100].index)
if 'passenger_count' in df.columns:
    problematic_rows.update(df[df['passenger_count'] == 0].index)
    problematic_rows.update(df[df['passenger_count'] > 6].index)
if 'trip_duration' in df.columns:
    problematic_rows.update(df[df['trip_duration'] < 0].index)

print(f"\nGesamtanzahl Zeilen mit identifizierten Problemen: {len(problematic_rows):,}")
print(f"Prozentsatz problematischer Daten: {(len(problematic_rows) / total_rows) * 100:.2f}%")
```
